```python
def cek_angka(a,b,c):

    if a!=b and a!=c and b!=c:

        if a+b==c or a+c==b or b+c==a:
            return True

    return False


print(cek_angka(2,3,5))
```

```python
def cek_digit_belakang(a,b,c):

    a = a % 10
    b = b % 10
    c = c % 10

    if a==b or a==c or b==c:
        return True
    else:
        return False


x = int(input("Masukkan angka 1: "))
y = int(input("Masukkan angka 2: "))
z = int(input("Masukkan angka 3: "))

print(cek_digit_belakang(x,y,z))
```

```python
c_to_f = lambda c : (9/5)*c + 32
c_to_r = lambda c : 0.8*c

print("Fahrenheit:", c_to_f(100))
print("Reamur:", c_to_r(80))
print("Fahrenheit:", c_to_f(0))
```
