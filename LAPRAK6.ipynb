```python
def perkalian(a, b):
    hasil = 0
    for i in range(a):
        hasil = hasil + b
    return hasil

print(perkalian(6,5))
print(perkalian(7,10))
```

```python
def ganjil(bawah, atas):

    if bawah < atas:
        for i in range(bawah, atas+1):
            if i % 2 == 1:
                print(i, end=" ")

    else:
        for i in range(bawah, atas-1, -1):
            if i % 2 == 1:
                print(i, end=" ")

ganjil(10,30)
print()
ganjil(97,82)
```

```python
jumlah = int(input("Jumlah mata kuliah: "))

total = 0

for i in range(jumlah):

    nilai = input("Masukkan nilai (A/B/C/D): ")

    if nilai == "A":
        bobot = 4
    elif nilai == "B":
        bobot = 3
    elif nilai == "C":
        bobot = 2
    elif nilai == "D":
        bobot = 1

    total = total + bobot

ips = total / jumlah

print("IPS =", ips)
```
