```python
upah_per_jam = float(input("Masukkan upah per jam: "))
jam_per_minggu = float(input("Masukkan jam kerja per minggu: "))

lama_kerja = 5

penghasilan_awal = upah_per_jam * jam_per_minggu * lama_kerja

pajak = 0.14 * penghasilan_awal
penghasilan_setelah_pajak = penghasilan_awal - pajak

biaya_pakaian = 0.10 * penghasilan_setelah_pajak
biaya_alat_tulis = 0.01 * penghasilan_setelah_pajak

saldo_sisa = penghasilan_setelah_pajak - biaya_pakaian - biaya_alat_tulis

total_donasi = 0.25 * saldo_sisa

donasi_anak_yatim = 0.30 * total_donasi
donasi_kaum_dhuafa = total_donasi - donasi_anak_yatim

print("\n===== RINCIAN PERHITUNGAN =====")
print("Penghasilan awal          :", penghasilan_awal)
print("Pajak (14%)               :", pajak)
print("Penghasilan setelah pajak :", penghasilan_setelah_pajak)
print("Biaya pakaian (10%)       :", biaya_pakaian)
print("Biaya alat tulis (1%)     :", biaya_alat_tulis)
print("Saldo sisa                :", saldo_sisa)
print("Total donasi (25%)        :", total_donasi)
print("Donasi anak yatim (30%)   :", donasi_anak_yatim)
print("Donasi kaum dhuafa        :", donasi_kaum_dhuafa)
```
