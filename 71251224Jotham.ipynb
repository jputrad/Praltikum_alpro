```python
def cek_anagram(kata1, kata2):
  
    string1 = sorted(kata1.lower().replace(" ", ""))
    string2 = sorted(kata2.lower().replace(" ", ""))
    
    if string1 == string2:
        return f"{kata1} dan {kata2} adalah anagram."
    else:
        return f"{kata1} dan {kata2} bukan anagram."


print(cek_anagram("mata", "atma"))
print(cek_anagram("buku", "kubu"))
```

```python
def hitung_frekuensi(kalimat, kata_cari):
    
    kalimat_bersih = kalimat.lower().replace(".", "").replace(",", "")
    daftar_kata = kalimat_bersih.split()
    
    jumlah = daftar_kata.count(kata_cari.lower())
    return f"{kata_cari} ada {jumlah} buah"


teks = "Saya mau makan. Makan itu wajib. Mau siang atau malam saya wajib makan"
print(hitung_frekuensi(teks, "makan"))
```

```python
def hapus_spasi_lebih(kalimat):
    
    hasil = " ".join(kalimat.split())
    return hasil


input_teks = "saya  tidak  suka  memancing  ikan   "
print(f"Output: '{hapus_spasi_lebih(input_teks)}'")
```

```python
def cari_pendek_panjang(kalimat):
    daftar_kata = kalimat.split()
    
    terpendek = min(daftar_kata, key=len)
    terpanjang = max(daftar_kata, key=len)
    
    return f"terpendek: {terpendek}, terpanjang: {terpanjang}"


teks_latihan = "red snakes and a black frog in the pool"
print(cari_pendek_panjang(teks_latihan))
```

```python
import re
from datetime import datetime

def olah_tanggal(teks):
    
    pola = r"(\d{4})-(\d{2})-(\d{2})"
    sekarang = datetime.now()
    
    temukan = re.findall(pola, teks)
    
    for tgl in temukan:
        
        str_tgl = f"{tgl[0]}-{tgl[1]}-{tgl[2]}"
        obj_tgl = datetime.strptime(str_tgl, "%Y-%m-%d")
        
       
        selisih = abs((sekarang - obj_tgl).days)
        
     
        print(f"{tgl[2]}-{tgl[1]}-{tgl[0]} selisih {selisih} hari")


data_teks = "Pada tanggal 1945-08-17 Indonesia merdeka. Pangeran Diponegoro (TL: 1785-11-11)"
olah_tanggal(data_teks)
```

```python
import re
import random
import string

def kelola_email(teks):
    
    emails = re.findall(r"(\w+)@\S+", teks)
    
    for user in emails:
      
        pool = string.ascii_letters + string.digits
        password = "".join(random.choice(pool) for _ in range(8))
        
     
        email_lengkap = f"{user}@..." # Ilustrasi
        print(f"username: {user} , password: {password}")

data_email = """
anton@mail.com dimiliki oleh antonius
budi@gmail.co.id dimiliki oleh budi anwari
"""
kelola_email(data_email)
```
